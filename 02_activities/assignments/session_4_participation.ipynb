## Seaborn Line Plot

Basic line plot showing relationship between variables.

```python
# Create a basic line plot
sns.lineplot(data=tips,           # df
             x='total_bill',      # X-axis 
             y='tip')             # Y-axis 

plt.title('Tips vs. Total Bill')
plt.show()
```

several built-in styles to quickly change plot appearance.

```python
sns.set_style('whitegrid')
sns.lineplot(data=tips, x='total_bill', y='tip')
plt.title('Tips vs. Total Bill (whitegrid style)')
plt.show()
```

Seaborn plots return matplotlib axes objects, allowing full customization.

```python
# Create figure with specific size
fig = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip',
                        color='hotpink',     
                        linestyle='--',       
                        linewidth=3,         
                        marker='o',          
                        markerfacecolor='indigo')  

tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')

plt.tight_layout()
plt.show()
```

Seaborn and matplotlib work together
- Use Seaborn for quick, beautiful statistical plots
- Use matplotlib methods for fine-tuned customization

Adding Multiple Visual Elements

```python
# Create a more complex customized plot
fig, ax = plt.subplots(figsize=(10, 5))

tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip',
                        color='hotpink',
                        linestyle='--',
                        linewidth=3,
                        marker='o',
                        markerfacecolor='indigo',
                        ax=ax)  # Attach to specific axis

tipgraph.set(title='Restaurant Tips Analysis',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')

# font
ax.text(40, 8, 'Higher bills generally\ncorrelate with higher tips',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()
```


Scatter plots

```python
# Create scatter plot with categorical coloring
tipgraph = sns.scatterplot(data=tips,
                           x='total_bill',
                           y='tip',
                           style='time',        # Different markers for Lunch/Dinner
                           hue='day',           
                           palette=['purple', 'hotpink', 'deepskyblue', 'yellowgreen'],
                           s=100)               

tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')

plt.tight_layout()
plt.show()
```


- `hue`: Color points by categorical variable
- `style`: Change marker shape by categorical variable
- `palette`: Custom color scheme


Pairplot creates a matrix showing relationships between all numerical variables.

```python
# Create pairplot for all numerical columns
sns.pairplot(
    tips,
    vars=['total_bill', 'tip', 'size'],  # Select specific columns
    hue='time',                          # Color by meal time
    palette=['purple', 'yellowgreen'],
    diag_kind='kde',                     # Use KDE for diagonal plots
    plot_kws={'alpha': 0.6}             # Transparency for overlapping points
)

plt.suptitle('Pairwise Relationships in Tips Dataset', y=1.02)
plt.tight_layout()
plt.show()
```

 Relationship Plots

```python
# Create a relational plot using Seaborn
daysgraph = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",        
    col="day",     
    col_wrap=2,         
    kind="scatter",     
    palette="day",
    height=3,            
    aspect=1.2      
)

daysgraph.set_axis_labels("Total Bill ($)", "Tip ($)")
daysgraph.set_titles("{col_name}")
plt.tight_layout()
plt.show()
```

## Part 2: Plotly - Interactive Visualizations
### Key Difference from Matplotlib/Seaborn:
- Matplotlib/Seaborn: Static images
- Plotly: Interactive, dynamic visualizations

```python
# Import Plotly
import plotly.graph_objects as go  # 'go' stands for 'graph objects'
```

### Basic Plotly Scatter Plot

```python
# Create a Plotly figure object
graph = go.Figure()

# Add scatter trace
graph.add_trace(
    go.Scatter(
        x=tips['total_bill'],
        y=tips['tip'],
        mode='markers',         
        marker=dict(
            size=8,
            color='deepskyblue',
            line=dict(width=1, color='navy')  
        ),
        name='Tips Data',       
        hovertemplate='Bill: $%{x:.2f}<br>Tip: $%{y:.2f}<extra></extra>'  
    )
)

# Update layout
graph.update_layout(
    title='Interactive Tips Visualization',
    xaxis_title='Total Bill ($)',
    yaxis_title='Tip ($)',
    hovermode='closest',
    template='plotly_white'  
)

# Show 
graph.show()
```

### Multiple Traces in Plotly

```python
graph = go.Figure()
day_colors = {
    'Thur': 'lightcoral',
    'Fri': 'gold',
    'Sat': 'lightgreen',
    'Sun': 'lightskyblue'
}

# Add a separate trace for each day
for day in tips['day'].unique():
    day_data = tips[tips['day'] == day]
    
    graph.add_trace(
        go.Scatter(
            x=day_data['total_bill'],
            y=day_data['tip'],
            mode='markers',
            marker=dict(
                size=10,
                color=day_colors.get(day, 'gray'),
                line=dict(width=1, color='darkgray')
            ),
            name=day,  # This appears in legend
            hovertemplate=f'<b>{day}</b><br>Bill: $%{{x:.2f}}<br>Tip: $%{{y:.2f}}<extra></extra>'
        )
    )

graph.update_layout(
    title='Tips by Day of Week (Interactive)',
    xaxis_title='Total Bill ($)',
    yaxis_title='Tip ($)',
    hovermode='closest',
    template='plotly_white',
    legend=dict(
        title='Day of Week',
        yanchor='top',
        y=0.99,
        xanchor='left',
        x=0.01
    )
)

graph.show()
```

## Part 3: WordCloud - Text Visualization

WordCloud

### Load Data with Text

```python
# Import WordCloud library
from wordcloud import WordCloud
# Load a dataset with text (using seaborn's quotes dataset)
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/quotes.csv')

print("Dataset shape:", df.shape)
print("\nFirst few quotes:")
print(df.head())
```

### Create Basic WordCloud

```python
# Join all text from the quote column into one long string
all_text = ' '.join(df['quote'].astype(str))

wordcloud = WordCloud(
    width=800,              
    height=400,            
    background_color='white',  
    colormap='viridis',     
    max_words=100,          
    relative_scaling=0.5,   
    min_font_size=10       
).generate(all_text)

# Display
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Remove axes
plt.title('Word Cloud from Quotes Dataset', fontsize=16, pad=20)
plt.tight_layout()
plt.show()
```

### WordCloud with Custom Stop Words

```python

from wordcloud import STOPWORDS

# default stop words
custom_stopwords = set(STOPWORDS)
custom_stopwords.update(['said', 'will', 'one', 'two', 'now', 'well'])

# custom stop words
wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color='white',
    colormap='Set2',
    stopwords=custom_stopwords,  # Use custom stop words!!
    max_words=100,
    relative_scaling=0.5
).generate(all_text)

plt.figure(figsize=(14, 7))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Word Cloud with Custom Stop Words Removed', fontsize=16, pad=20)
plt.tight_layout()
plt.show()
```

## Part 4: Venn Diagrams - Set Visualization

### Basic Venn Diagram with Two Sets

```python
# Import Venn diagram functions
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

# Define two sets
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

print("Set A (Fruits):", A)
print("Set B (Pies):", B)
print("Intersection (A ∩ B):", A & B)
print("Only in A (A - B):", A - B)
print("Only in B (B - A):", B - A)
```

```python
# Create Venn diagram
plt.figure(figsize=(8, 6))

diagram = venn2_unweighted(
    [A, B],
    set_labels=('Fruits', 'Pies'),     # Labels for each set
    set_colors=("blue", "red"),        # Colors for each circle
    alpha=0.5                          # Transparency
)

diagram.get_label_by_id('10').set_text('\n'.join(A - B))
diagram.get_label_by_id('11').set_text('\n'.join(A & B))
diagram.get_label_by_id('01').set_text('\n'.join(B - A))

plt.title('Venn Diagram: Fruits vs. Pies', fontsize=14, pad=20)
plt.tight_layout()
plt.show()
```

**Understanding index:**
- `'10'`: Only in first set (left side only)
- `'01'`: Only in second set (right side only)
- `'11'`: In both sets (intersection/overlap)

### Enhanced Venn Diagram with Circles and Styling

```python
# Create a more professional-looking Venn diagram
plt.figure(figsize=(10, 7))

diagram = venn2_unweighted(
    [A, B],
    set_labels=('Fruits', 'Pies'),
    set_colors=("lightblue", "lightcoral"),
    alpha=0.6
)

# contour
venn2_circles([A, B], linewidth=2, linestyle='--', color='black')

# labels
diagram.get_label_by_id('10').set_text('\n'.join(A - B))
diagram.get_label_by_id('11').set_text('\n'.join(A & B))
diagram.get_label_by_id('01').set_text('\n'.join(B - A))
```

## Summary

### 1. Seaborn:
- High-level statistical visualization library
- Built on matplotlib with better defaults
- Easy categorical plotting with `hue`, `style`, `size`
- Powerful multi-plot functions: `pairplot`, `relplot`

### 2. Plotly:
- Interactive, web-based visualizations
- Great for presentations and dashboards

### 3. WordCloud:
- Visualize text data frequency
- Larger words = more frequent
- Useful for survey responses, social media, documents

### 4. Venn Diagrams:
- Show set relationships and overlaps
- Use `venn2` for two sets, `venn3` for three sets
- Label regions with actual items or counts

## When to Use Each Tool:

| Tool | Best For |
|------|----------|
| **Matplotlib** | Full control, publication-quality static plots |
| **Seaborn** | Quick statistical exploration, beautiful defaults |
| **Plotly** | Interactive exploration, dashboards, presentations |
| **WordCloud** | Text analysis, survey responses, keyword visualization |
| **Venn** | Set comparisons, group overlaps |
